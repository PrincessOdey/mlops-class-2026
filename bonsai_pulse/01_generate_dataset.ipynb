```python
import pandas as pd
import numpy as np

np.random.seed(42)

def generate_row():
    cause = np.random.choice(
        ["healthy", "overwatering", "underwatering", "pest", "low_light"],
        p=[0.25, 0.2, 0.2, 0.15, 0.2]
    )

    if cause == "healthy":
        leaf_color = np.random.choice(["green", "pale_green"], p=[0.85, 0.15])
        soil_moisture = np.random.normal(50, 12)
        light_hours = np.random.normal(6.5, 1.5)
        days_since_watered = np.random.randint(1, 6)
        pest_sighted = np.random.choice([0, 1], p=[0.95, 0.05])

    elif cause == "overwatering":
        leaf_color = np.random.choice(["yellow", "brown", "green"], p=[0.55, 0.25, 0.2])
        soil_moisture = np.random.normal(80, 15)
        light_hours = np.random.normal(6, 2)
        days_since_watered = np.random.randint(0, 3)
        pest_sighted = np.random.choice([0, 1], p=[0.9, 0.1])

    elif cause == "underwatering":
        leaf_color = np.random.choice(["brown", "yellow", "green"], p=[0.5, 0.35, 0.15])
        soil_moisture = np.random.normal(15, 12)
        light_hours = np.random.normal(6, 2)
        days_since_watered = np.random.randint(5, 14)
        pest_sighted = np.random.choice([0, 1], p=[0.92, 0.08])

    elif cause == "pest":
        leaf_color = np.random.choice(["yellow", "green", "brown"], p=[0.45, 0.4, 0.15])
        soil_moisture = np.random.normal(45, 15)
        light_hours = np.random.normal(6, 2)
        days_since_watered = np.random.randint(1, 8)
        pest_sighted = np.random.choice([0, 1], p=[0.25, 0.75])

    else:  # low_light
        leaf_color = np.random.choice(["pale_green", "yellow", "green"], p=[0.5, 0.3, 0.2])
        soil_moisture = np.random.normal(45, 15)
        light_hours = np.random.normal(2, 1.2)
        days_since_watered = np.random.randint(1, 8)
        pest_sighted = np.random.choice([0, 1], p=[0.93, 0.07])

    soil_moisture = float(np.clip(soil_moisture, 0, 100))
    light_hours = float(np.clip(light_hours, 0, 12))

    return {
        "leaf_color": leaf_color,
        "soil_moisture_pct": round(soil_moisture, 1),
        "light_hours_per_day": round(light_hours, 1),
        "days_since_watered": days_since_watered,
        "pest_sighted": pest_sighted,
        "likely_cause": cause
    }
```

```python
rows = [generate_row() for _ in range(300)]
df = pd.DataFrame(rows)
df.head(10)
```

```python
   import os
   os.makedirs("data", exist_ok=True)
   df.to_csv("data/bonsai_symptoms.csv", index=False)
   print("Saved", len(df), "rows to data/bonsai_symptoms.csv")
```
